```python
def check_scholarship(score, documents, attendance, income):

    if score < 70:
        return "Not Eligible"

    if documents == False:
        return "Not Eligible"

    if attendance < 75:
        return "Not Eligible"

    if income > 50000:
        return "Not Eligible"

    return "Eligible"


# Test Case 1: All conditions pass
print("Test 1:", check_scholarship(85, True, 90, 40000))

# Test Case 2: Attendance fails
print("Test 2:", check_scholarship(85, True, 60, 40000))

# Test Case 3: Family income fails
print("Test 3:", check_scholarship(85, True, 90, 60000))

# Test Case 4: Score is exactly 70
print("Test 4:", check_scholarship(70, True, 80, 40000))
```
