```python
import math

scores = [3, 5, 2, 9, 3, 5, 2, 9]

tree_depth = int(math.log2(len(scores)))

counter = [0]

def minimax(depth, index, is_max, scores, tree_depth, counter):

    if depth == tree_depth:
        counter[0] += 1
        return scores[index]

    if is_max:
        left = minimax(depth + 1, index * 2, False,
                       scores, tree_depth, counter)

        right = minimax(depth + 1, index * 2 + 1, False,
                        scores, tree_depth, counter)

        return max(left, right)

    # MIN player
    else:
        left = minimax(depth + 1, index * 2, True,
                       scores, tree_depth, counter)

        right = minimax(depth + 1, index * 2 + 1, True,
                        scores, tree_depth, counter)

        return min(left, right)


result = minimax(0, 0, True, scores, tree_depth, counter)

print("Tree Depth:", tree_depth)
print("Minimax Value:", result)
print("Leaves Evaluated:", counter[0])
```

```python
import math


scores = [3, 5, 2, 9, 3, 5, 2, 9]


tree_depth = int(math.log2(len(scores)))

counter = [0]
def alphabeta(depth, index, is_max, scores, tree_depth,
              alpha, beta, counter):
    if depth == tree_depth:
        counter[0] += 1
        return scores[index]

    if is_max:
        value = float('-inf')

        for i in range(2):

            value = max(
                value,
                alphabeta(
                    depth + 1,index * 2 + i, False, scores, tree_depth, alpha, beta, counter  ) )

            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:
        value = float('inf')

        for i in range(2):

            value = min(
                value,
                alphabeta( depth + 1, index * 2 + i, True, scores, tree_depth, alpha, beta, counter ))

            beta = min(beta, value)

            if alpha >= beta:
                break

        return value

result = alphabeta(
    0, 0, True, scores, tree_depth, float('-inf'), float('inf'), counter)

print("Tree Depth:", tree_depth)
print("Alpha-Beta Value:", result)
print("Leaves Evaluated:", counter[0])
```
