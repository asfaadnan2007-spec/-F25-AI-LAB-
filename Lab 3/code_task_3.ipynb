```python
import heapq
weighted_graph = {
    "A": [("B", 1), ("C", 6)],
    "B": [("D", 1), ("E", 1)],
    "C": [("F", 1), ("G", 6)],
    "D": [],
    "E": [("G", 1)],
    "F": [],
    "G": []
}

def ucs(start, goal):
    priority_queue = [(0, start, [start])]
    visited = set()
    processing_order = []

    while priority_queue:
        cost, node, path = heapq.heappop(priority_queue)

        if node in visited:
            continue

        visited.add(node)
        processing_order.append(node)

        if node == goal:
            return path, cost, processing_order

        for neighbor, weight in weighted_graph[node]:
            if neighbor not in visited:
                heapq.heappush(
                    priority_queue,
                    (cost + weight, neighbor, path + [neighbor])
                )

    return None, float("inf"), processing_order

path, cost, order = ucs("A", "G")

print("UCS Processing Order:", order)
print("UCS Shortest Path:", path)
print("Total Cost:", cost)
```

```python

from collections import deque

def bfs_weighted(start, goal):
    queue = deque([(start, [start])])
    visited = {start}
    processing_order = []

    while queue:
        node, path = queue.popleft()
        processing_order.append(node)

        if node == goal:
            return path, processing_order

        for neighbor, weight in weighted_graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None, processing_order


bfs_path, bfs_order = bfs_weighted("A", "G")

bfs_cost = sum(
    next(weight for neighbor, weight in weighted_graph[node]
         if neighbor == bfs_path[i + 1])
    for i, node in enumerate(bfs_path[:-1])
)

print("BFS Processing Order:", bfs_order)
print("BFS Path:", bfs_path)
print("BFS Total Cost:", bfs_cost)

print("\nUCS Processing Order:", order)
print("UCS Path:", path)
print("UCS Total Cost:", cost)
```

```python

# Independently calculate UCS path cost

ucs_verified_cost = 0
path =[]
cost = 0
for i in range(len(path) - 1):
    current = path[i]
    next_node = path[i + 1]

    for neighbor, weight in weighted_graph[current]:
        if neighbor == next_node:
            ucs_verified_cost += weight

print("UCS Path:", path)
print("Calculated Cost:", ucs_verified_cost)
print("Original UCS Cost:", cost)

if ucs_verified_cost == cost:
    print("Cost Verification: PASSED")
else:
    print("Cost Verification: FAILED")
```
