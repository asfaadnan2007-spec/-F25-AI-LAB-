```python
import heapq

def run_search(graph, h, start, goal, use_heuristic=True):
    # Priority queue stores tuples of (f, g, node, path)
    initial_h = h[start] if use_heuristic else 0
    pq = [(initial_h, 0, start, [start])]
    best = {start: 0}
    expanded_nodes = []
    
    while pq:
        f, g, node, path = heapq.heappop(pq)
        
        if g != best.get(node, float("inf")):
            continue
            
        # Goal processing logged, but goal is excluded from expansion count
        if node == goal:
            return g, path, expanded_nodes
            
        # Count node as expanded when its successors are examined
        expanded_nodes.append(node)
        
        for nxt, w in graph.get(node, []):
            ng = g + w
            if ng < best.get(nxt, float("inf")):
                best[nxt] = ng
                nh = h[nxt] if use_heuristic else 0
                heapq.heappush(pq, (ng + nh, ng, nxt, path + [nxt]))
                
    return None, [], expanded_nodes

# Graph Edges & Weights
graph = {
    "S": [("A", 1), ("B", 4)], 
    "A": [("C", 2)], 
    "B": [("G", 5)], 
    "C": [("G", 3)], 
    "G": []
}

# Heuristic Maps
h_astar = {"S": 5, "A": 4, "B": 4, "C": 2, "G": 0}  # Supplied Heuristic
h_ucs = {"S": 0, "A": 0, "B": 0, "C": 0, "G": 0}    # Zero-Heuristic (UCS)

if __name__ == "__main__":
    # Run A*
    cost_a, path_a, exp_a = run_search(graph, h_astar, "S", "G", use_heuristic=True)
    
    # Run UCS
    cost_u, path_u, exp_u = run_search(graph, h_ucs, "S", "G", use_heuristic=False)

    print("... TASK 2 RESULTS ....")
    print(f"A*  -> Path: {path_a} | Cost: {cost_a} | Expanded: {exp_a} (Count: {len(exp_a)})")
    print(f"UCS -> Path: {path_u} | Cost: {cost_u} | Expanded: {exp_u} (Count: {len(exp_u)})")
```

```python
# Explanation of Differences:Counting Convention: Goal node G is logged upon reach but excluded from the expansion count
# because its successors are not examined.   Efficiency Difference: Both algorithms find the optimal path with cost 6.
# However, A* explores 3 nodes while UCS explores 4 nodes because A* heuristic guides it away from expanding node B prematurely.  
```
