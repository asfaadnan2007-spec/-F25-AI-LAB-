```python
import heapq

def run_search(graph, h, start, goal, use_heuristic=True):
    # Priority queue stores tuples of (f, g, node, path)
    initial_h = h[start] if use_heuristic else 0
    pq = [(initial_h, 0, start, [start])]
    best = {start: 0}
    expanded_nodes = []
    
    while pq:
        f, g, node, path = heapq.heappop(pq)
        
        # Ignore stale queue entries
        if g != best.get(node, float("inf")):
            continue
            
        # Goal processing logged, but goal is excluded from expansion count
        if node == goal:
            return g, path, expanded_nodes
            
        # Count node as expanded when its successors are examined
        expanded_nodes.append(node)
        
        for nxt, w in graph.get(node, []):
            ng = g + w
            if ng < best.get(nxt, float("inf")):
                best[nxt] = ng
                nh = h[nxt] if use_heuristic else 0
                heapq.heappush(pq, (ng + nh, ng, nxt, path + [nxt]))
                
    return None, [], expanded_nodes

# Define Graph Edges & Weights
graph = {
    "S": [("A", 1), ("B", 4)], 
    "A": [("C", 2)], 
    "B": [("G", 5)], 
    "C": [("G", 3)], 
    "G": []
}

# Original heuristic dictionary
h_astar = {"S": 5, "A": 4, "B": 4, "C": 2, "G": 0}

# Modified heuristic dictionary for Task 3 (Overestimate h(C) = 10)
h_overestimate = {"S": 5, "A": 4, "B": 4, "C": 10, "G": 0}

if __name__ == "__main__":
    print("=== TASK 3: OVERESTIMATE EXPERIMENT ===")
    
    # 1. Run Original (Admissible)
    cost_orig, path_orig, exp_orig = run_search(graph, h_astar, "S", "G", use_heuristic=True)
    print(f"Original A* (h(C) = 2):")
    print(f"  Path:     {path_orig}")
    print(f"  Cost:     {cost_orig}")
    print(f"  Expanded: {exp_orig} (Count: {len(exp_orig)})\n")
    
    # 2. Run Overestimate (Inadmissible)
    cost_over, path_over, exp_over = run_search(graph, h_overestimate, "S", "G", use_heuristic=True)
    print(f"Modified A* (h(C) = 10):")
    print(f"  Path:     {path_over}")
    print(f"  Cost:     {cost_over}")
    print(f"  Expanded: {exp_over} (Count: {len(exp_over)})\n")
    
    # 3. Summary of Changes
    print("=== COMPARISON SUMMARY ===")
    print(f"Did Path Change?  {'Yes' if path_orig != path_over else 'No'} ({path_orig} -> {path_over})")
    print(f"Did Cost Change?  {'Yes' if cost_orig != cost_over else 'No'} ({cost_orig} -> {cost_over})")
    print(f"Did Count Change? {'Yes' if len(exp_orig) != len(exp_over) else 'No'} ({len(exp_orig)} -> {len(exp_over)})")
```
